```python
from dotenv import load_dotenv
import os 
from ibm_watsonx_ai import APIClient
from ibm_watsonx_ai import Credentials
from ibm_watsonx_ai.foundation_models import ModelInference
```

```python
# .env 내용 가져오기
load_dotenv()

apiKey = os.getenv("WATSONX_API_KEY")
project_id = os.getenv("WATSONX_PROJECT_ID")
watsonx_ai_url = os.getenv("WATSONX_URL")
```

```python
credentials = Credentials(
               url = f"{watsonx_ai_url}",
               api_key = f"{apiKey}",
                  )
api_client = APIClient(credentials)

api_client.foundation_models.TextModels.show()
```

```python
credentials = Credentials(
   url = f"{watsonx_ai_url}",
   api_key = f"{apiKey}",
)
client = APIClient(credentials)

model = ModelInference(
    model_id="ibm/granite-4-h-small",
    api_client=client,
    project_id=f"{project_id}",
    params = {
        "max_tokens": 100
}
)
# 사용자 프롬프트
prompt = 'How far is Paris from Bangalore?'
messages = [
    # 시스템 프롬프트
    # {"role":"system", "content":"You are a helpful assistant."},
    {"role":"user", "content":prompt},

]

generated_response = model.chat(messages=messages)

print(generated_response)
print(generated_response['choices'][0]['message']['content'])
```

```python
credentials = Credentials(
   url = f"{watsonx_ai_url}",
   api_key = f"{apiKey}",
)
client = APIClient(credentials)

model = ModelInference(
    model_id="ibm/granite-4-h-small",
    api_client=client,
    project_id=f"{project_id}",
    params = {
        "max_tokens": 2000,
        "temperature": 1
}
)
# 사용자 프롬프트
user_prompt = """무서운 이야기 만들어줄래?
"""

system_prompt = """
너는 유명한 공포 소설 작가야.
"""

messages = [
    # 시스템 프롬프트
    {"role":"system", "content":system_prompt},
    {"role":"user", "content":user_prompt},

]

# generated_response = model.chat(messages=messages)

# print(generated_response)
# print(generated_response['choices'][0]['message']['content'])


generated_response_stream = model.chat_stream(messages=messages) # 타자 치듯이 동작하는 챗 모델
for chunk in generated_response_stream:
    if chunk['choices']:
        print(chunk['choices'][0]['delta'].get('content',''),end="", flush=True)
```
