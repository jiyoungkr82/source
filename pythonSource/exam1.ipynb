### 조건문
- 조건에 맞는 상황을 판단한 후 수행하도록 처리
- ex) 현재 변수에 값이 5라면, dict1 키가 name이 포함되어 있다면
```
if 조건문: 
    조건이 참일 때 수행할 문장
```

````python
if 'name' in dict1:
    print(dict1.get('name'))
else:
    print('None')
````

````python
# input()을 통해 받는 모든 것은 문자 타입
score = input("당신의 점수를 입력하세요.")
b = 90
# print(type(score)) => class str
# print(type(b)) => class int

# int 타입으로 변경
score = int(score) # 원본 변수에 변경된 값을 다시 할당해야 함
result = ''

if score >=90 :
    result = 'A'
elif score >=80:
    result = 'B'
elif score >=70:
    result = 'C'
else:
    result = 'D'

# print(f"당신의 성적은 {result} 입니다")
# print("당신의 성적은 %s 입니다" %result)

# 점수가 90 이상이고 평가 등급이 A 라면 합격입니다. 그외는 불합격입니다.
if score >=90 and result == 'A':
    print(f"당신의 학점은 {result}이고, 결과는 합격입니다.")
else:
    print(f"당신의 학점은 {result}이고, 결과는 불합격입니다.")



````
