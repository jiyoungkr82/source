````python
def add_many(*args):
    result = 0
    for i in args:
        result += i
    return result

print(add_many(3,4,5,6,2))
print(add_many(1,2,3,4))
print(add_many())
````

````python
def add_many(choice, *args):
    if choice == "mul":
        result = 1
        for i in args:
            result *= i
    elif choice == "add":
        result = 0
        for i in args:
            result += i
    return result


print(add_many("add", 3,4,5,6,3))
````

````python
# **kargs : dict 형식 
# *args : 입력의 개수가 지정되지 않았을 때
# 변수명 : 1:1 대응

def add_many(**kwargs):
    for k, v in kwargs.items():
        print(k,v)

add_many(name="kim", active="false")
````

### 클래스
- 함수, 변수를 포함한 형태
- 객체지향

````python
# 클래스 생성
class Calculator:
    def __init__(self):
        self.result = 0

    def add(self, num):
        self.result += num
        return self.result

    def sub(self, num):
        self.result -= num
        return self.result
````

````python
# 사용 - 객체 생성
calc = Calculator()

# 함수(메소드) 호출
print(calc.add(5))
print(calc.sub(5))
````

````python
# 사칙연산 클래스
class FourCal:
    def __init__(self, first, second):
        self.first = first
        self.second = second

    # 메서드
    def add(self):
        return self.first + self.second

    def sub(self):
        return self.first - self.second
    
    def mul(self):
        return self.first * self.second
    
    def div(self):
        return self.first // self.second
    
    def __str__(self):
        return f"{self.first}, {self.second}"
````

````python
# __init__() 자동으로 호출됨
fcal = FourCal(5,3)

# 사칙연산
print(fcal.add())
print(fcal.sub())
print(fcal.mul())
print(fcal.div())
print(fcal)
````

````python
# 상속: 

class MoreFourCal(FourCal):
    def pow(self):
        return self.first ** self.second
````

````python
more = MoreFourCal(4,2)

print(more.pow())
````

````python
# 오버라이딩 : 부모 클래스 메소드 재정의

four1 = FourCal(4,2)
print(four1.div())
````

````python
class SafeFourCal(FourCal):
    def div(self):
        if self.second == 0:
            return 0
        else:
            return self.first / self.second
    
````

````python
four1 = SafeFourCal(4,0)
four1.div()
````

````python
# 나만의 오류 작성하기

class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명입니다."

def say_nick(nick):
    if nick == "바보":
        raise MyError()
    print(nick)
````
